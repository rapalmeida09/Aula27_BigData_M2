```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### Obtendo dados

```python
ENDERECO_DADOS = 'https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv'
print('Obtendo dados de ocorrências...')

df_ocorrencias = pd.read_csv(ENDERECO_DADOS, sep=';', encoding='iso-8859-1')
print(df_ocorrencias.head())

print('iniciando e delimitação das variaveis e totalizando')
df_ocorrencias = df_ocorrencias[['cisp', 'roubo_veiculo', 'recuperacao_veiculos']]

df_total_veiculos = df_ocorrencias.groupby('cisp', as_index=False)[['roubo_veiculo', 'recuperacao_veiculos']].sum()
print(df_total_veiculos.head())
```

### Excluir uma parte dos valores extremos

```python
print('analisando dados...')

df_total_veiculos_cut = (
    df_total_veiculos[
        df_total_veiculos['roubo_veiculo'] < np.percentile(
            df_total_veiculos['roubo_veiculo'], 95
        )
    ]
)

df_total_veiculos_cut = (
    df_total_veiculos_cut[
        df_total_veiculos_cut['recuperacao_veiculos'] < np.percentile(
            df_total_veiculos_cut['recuperacao_veiculos'], 95
        )
    ]
)
```

### Trasformando em Array

```python
array_roubo_veiculo = np.array(df_total_veiculos_cut['roubo_veiculo'])
array_recuperacao_veiculos = np.array(df_total_veiculos_cut['recuperacao_veiculos'])
```

### Calculando a Correlação de Pearson

```python

correlacao = np.corrcoef(array_roubo_veiculo, array_recuperacao_veiculos)[0, 1]
print(f'Correlação: {correlacao:.1f}')
```
